```python
from sys import displayhook

import networkx as nx
import matplotlib.pyplot as plt
import os
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
import numpy as np

csv_path = os.path.join("data", "vuelos_10k.csv")
df = pd.read_csv(csv_path)

# display(len(df['Destino']))
df_copy = df.copy()

df_copy = df_copy.dropna()
# df_copy = df_copy['Destino'].dropna()

len(df_copy)

df_copy
```

```python
df_copy['Destino'] = df_copy['Destino'].apply(lambda x: str(x).upper())
df_copy['Origen'] = df_copy['Origen'].apply(lambda x: str(x).upper())

df_copy
```

```python
display(df_copy[df_copy['Origen'] == df_copy['Destino']])
print(len(df_copy[df_copy['Origen'] == df_copy['Destino']]))

df_copy = df_copy[df_copy['Origen'] != df_copy['Destino']]
print(len(df_copy[df_copy['Origen'] == df_copy['Destino']]))
```

```python
df_copy = df_copy[df_copy['Distancia_km'] > 0]
df_copy
```

```python
print(len(df_copy))

df_copy = df_copy.drop_duplicates(subset=['Destino', 'Origen'], keep='last')

print(len(df_copy))
```

```python
grafo = nx.from_pandas_edgelist(df_copy,
                        source='Destino',
                        target='Origen',
                        )

print("Aristas: ", len(grafo.degree))
print("Nodos: ", len(grafo.edges))
```

```python
matriz_ady = nx.adjacency_matrix(grafo).todense()
print(matriz_ady.nbytes / 1000000)
print(df_copy.memory_usage(deep=True).sum() / 1000000)
```
