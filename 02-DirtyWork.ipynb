```python
import random

random_list = [random.random() for _ in range(5_000_000)]

print("Random list length:", len(random_list))
```

```python
### Desviacion Estandar

def average(random_list):
    avg = 0

    for i in random_list:
        avg += i

    avg /= len(random_list)
    print("Average:", avg)
    return avg

def standar_deviation(list, avg):
    std = 0

    for  i in list:
        std += (i - avg) ** 2
    std /= len(list)
    std = std ** 0.5

    print("Standard deviation:", std)
    return std

def variance(list, avg):
    variance = 0
    for i in list:
        variance += (i - avg) ** 2
    variance /= len(list)

    print("Variance:", variance)
    return variance

avg = average(random_list)
std = standar_deviation(random_list, avg)
```

```python
import numpy as np

np_array = np.array(random_list)

std = standar_deviation(np_array, avg)
```
